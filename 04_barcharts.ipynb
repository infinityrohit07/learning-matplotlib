## 2. **Grouped Bar Charts**

These are used when we want to compare different values of same category.

````python
months = ["Jan", "Feb", "Mar"]

laptop_sales = [40, 60, 50]
phone_sales = [30, 55, 65]

plt.bar(months, laptop_sales, color="red")
plt.bar(months, phone_sales, color="blue")

plt.show()
````

here both values for months category are overlapping. we need to seperate them. for this we use below way:

````python
import numpy as np # we imported numpy

months = ["Jan", "Feb", "Mar"]

laptop_sales = [40, 60, 50]
phone_sales = [70, 80, 65]

x = np.arange(len(months)) # we created a 1D array using numpy len(months) = 3 so array becomes [0,1,2]
width = 0.4 # we have taken a random value that will difference the position on x-axis for both

plt.bar(
    x, # so for this bar will start at [0,1,2]
    laptop_sales,
    width=width, # and it will end at [0.4, 1.4, 2.4]
    label="Laptop"
)

plt.bar(
    x + width, # the second one will start from [0.4, 1.4, 2.4]
    phone_sales,
    width=width, # the second one will end at [0.4 + 0.4, 1.4 + 0.4, 2.4 + 0.4]
    label="Phone"
)


# since we have passed x as category we won't have months name at x-axis [we will have 0.1, 0.5, 1.0, ...]
# so bring months name and set its position we used plt.xticks.
# for better understanding comment below one line and run the code and then again uncomment it and run the code and see the difference 
plt.xticks(x + width / 2, months) # other than that x+width/2 is done to place the month names exactly at middle of both the bars.

plt.title("Monthly Product Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.legend()  # we didn't discuss this in bar chart but we have done this in line plot and this is exactly same in barcharts too.
plt.show()
````

### **Sorting Bars**

Bar charts can be sorted based on their values.

Sorting is useful when we want to make it easier to see the order of categories from smallest to largest or largest to smallest.

For example:

- Laptop → 40
- Phone → 80
- Tablet → 30
- Watch → 60

Can be sorted as:

- Tablet → 30
- Laptop → 40
- Watch → 60
- Phone → 80

#### Important

When sorting a bar chart, the **category and its value must stay together**.

We should not sort only the values because that would break the relationship between each category and its value.

````python
import matplotlib.pyplot as plt

products = ["Laptop", "Phone", "Tablet", "Watch"]
sales = [40, 80, 30, 60]

bars = plt.bar(products, sales, edgecolor="black")

plt.bar_label(bars)

plt.title("Product Sales")
plt.xlabel("Product")
plt.ylabel("Sales")

plt.show()
````

```text
the above bar chart is unsorted as we can see. now let's learn how we can actually sort it.

now you must be thinking we can simply sort our sales and plot it? sales.sort() ?
the thing is if we do sales.sort() our sales becomes [30, 40, 60, 80]
but our products still remains ["Laptop", "Phone", "Tablet", "Watch"]
due to this we will lose the actual relationship between them.

so if this isn't the solution, what is?

#### **The solution is `zip()`**         
`zip()` combines corresponding elements from two sequences.

```python
zip(sales,products) # note we kept sales first not the product because when we will sort the sales will help to sort the products.
```
```text
the above python command creates:
(40, "Laptop")
(80, "Phone")
(30, "Tablet")
(60, "Watch")

````python
products = ["Laptop", "Phone", "Tablet", "Watch"]
sales = [40, 80, 30, 60]

data = zip(sales, products)

print(list(data))
````

note: the above code is only for explanation

now that we know `zip()` combined them, lets sort them using `sorted()`.        
`sorted()` returns items in sorted order.

so what we are gonna do is `sorted(zip(sales,products))`         
Since sales is placed first when we did `zip(sales,products)`, the sales values determine the sorting order.

````python
sorted_data = sorted(zip(sales,products))
print(list(sorted_data))
````

as we can see we have sorted it in combination. but now we have unzip it to have seperate values.                   
for this we will use `zip(*sorted_data)` and then store them into two seperate variables.

````python
sorted_sales, sorted_products = zip(*sorted_data)
print(sorted_sales)
print(sorted_products)
````

```text
Now that the values and categories are separated again, we can pass them to `plt.bar()`.
The bars will now appear from **smallest to largest**.

#### **Complete Code**

````python
import matplotlib.pyplot as plt

products = ["Laptop", "Phone", "Tablet", "Watch"]
sales = [40, 80, 30, 60]

# Sort from smallest to largest
sorted_data =  sorted(zip(sales,products))

# seperate the values
sorted_sales, sorted_products = zip(*sorted_data)

# create the bar chart
bars = plt.bar(sorted_products, sorted_sales, edgecolor="black")

plt.bar_label(bars)

plt.title("Products Sorted by Sales in ASC")
plt.xlabel("Product")
plt.ylabel("Sales")

plt.show()
````

### **Highlighting a Specific Bar**

Sometimes you want to make one particular bar stand out from the others. How do we do that?

Example if we want to highlight the max sale.

##### 1. Find its position first
```python
max_index = sales.index(max(sales))
```
```text\
max(sales) will give the max value.
sales.index(the_max_value) will give its position.
```

##### 2. Create a color list
```python
colors = ["gray"] * len(products)
```

##### 3. Change the color of the max_index
```python
colors[max_index] = "orange"
```

##### 4. Pass the colors to the color parameter while creating the bar
```python
plt.bar(products, sales, color=colors)

````python
import matplotlib.pyplot as plt

products = ["Laptop", "Phone", "Tablet", "Watch"]
sales = [40, 80, 30, 60]

# Find the index of the largest value
max_index = sales.index(max(sales))

# Give every bar the same color
colors = ["gray"] * len(products)

# Highlight the maximum bar
colors[max_index] = "orange"

bars = plt.bar(
    products,
    sales,
    color=colors,
    edgecolor="black"
)

plt.bar_label(bars)

plt.title("Product Sales")
plt.xlabel("Product")
plt.ylabel("Sales")

plt.show()
````

**Highlighting Both MAX and MIN Bar**

````python
import matplotlib.pyplot as plt

products = ["Laptop", "Phone", "Tablet", "Watch"]
sales = [40, 80, 30, 60]

# Find the index of the largest value
max_index = sales.index(max(sales))

# Find the index of the smallest value
min_index = sales.index(min(sales))

# Give every bar the same color
colors = ["gray"] * len(products)

# Highlight the maximum bar
colors[max_index] = "orange"

# Highlight the minimum bar
colors[min_index] = "yellow"

bars = plt.bar(
    products,
    sales,
    color=colors,
    edgecolor="black"
)

plt.bar_label(bars)

plt.title("Product Sales")
plt.xlabel("Product")
plt.ylabel("Sales")

plt.show()
````

### **Reference Line**

A **reference line** is a horizontal or vertical line added to a chart to represent an important value that we want to compare our data against.

It is commonly used to show:

- **Target** → compare values against a target
- **Average** → compare values against the average
- **Limit** → show a maximum or minimum allowed value
- **Benchmark** → compare values against a standard value

#### Horizontal Reference Line

For a bar chart, we commonly use `plt.axhline()` to add a horizontal reference line.

```python
plt.axhline(y=value)

````python
import matplotlib.pyplot as plt

products = ["Laptop", "Phone", "Tablet", "Watch"]
sales = [40, 80, 30, 60]

# The value we want to use as our target
target = 50

# Create the bar chart
bars = plt.bar(
    products,
    sales,
    edgecolor="black"
)

plt.bar_label(bars)

# Add a horizontal reference line at Y = 50
# y=target → determines where the line is placed
# linestyle="--" → makes the line dashed
# label="Target" → gives the line a name for the legend
plt.axhline(
    y=target,
    linestyle="--",
    label="Target"
)


plt.title("Product Sales")
plt.xlabel("Product")
plt.ylabel("Sales")

plt.legend()

plt.show()
````

### **Error Bars**

**Error bars** are used to show the **variation or uncertainty associated with a value**.

In a bar chart, the bar represents the main value, while the error bar shows how much variation or uncertainty is associated with that value.

For example:

```text
Value = 50
Error = 5

50 ± 5

````python
import matplotlib.pyplot as plt

products = ["Laptop", "Phone", "Tablet"]
sales = [50, 80, 40]

# How much variation/uncertainty is associated with each value
errors = [5, 8, 4]

# Create the bar chart
# yerr adds vertical error bars to each bar
plt.bar(
    products,
    sales,
    yerr=errors,
    capsize=5
)

plt.title("Sales with Error Bars")
plt.xlabel("Product")
plt.ylabel("Sales")

plt.show()
````
