## 4. **Scatter Plot**
```text
Scatter Plot are used to study the relationship between two numerical variables.

Example:
the amount of time a student studies have a relationship with their exam marks?

Marks
80 |                  ●
70 |              ●
60 |          ●
50 |       ●
40 |   ●  ●
30 | ●
   +----------------------
     1  2  3  4  5  6  7
          Hours studied

One x-value + one y-value = one observation/point.

```

**Basic Syntax:**

```python
plt.scatter(x, y)
```

````python
import matplotlib.pyplot as plt

# Number of hours each student studied
hours_studied = [1, 2, 3, 4, 5, 6, 7]

# Marks obtained by those students
marks = [35, 42, 48, 55, 63, 71, 78]

# Create a scatter plot
# x-axis → hours studied
# y-axis → marks
plt.scatter(hours_studied, marks)

# Display the plot
plt.show()
````

### **edgecolor**
```text
edgecolor controls the border around each scatter point.

````python
import matplotlib.pyplot as plt

# Number of hours each student studied
hours_studied = [1, 2, 3, 4, 5, 6, 7]

# Marks obtained by those students
marks = [35, 42, 48, 55, 63, 71, 78]

# Create the scatter plot
plt.scatter(
    hours_studied,
    marks,
    color="skyblue",       # Fill color of the points
    s=100,                 # Size of the points
    alpha=0.7,             # Transparency
    edgecolor="black"      # Border color of the points
)

plt.show()
````

### **linewidth**
```text
linewidth controls the width of the border of the points.

````python
import matplotlib.pyplot as plt

# Number of hours each student studied
hours_studied = [1, 2, 3, 4, 5, 6, 7]

# Marks obtained by those students
marks = [35, 42, 48, 55, 63, 71, 78]

# Create the scatter plot
plt.scatter(
    hours_studied,
    marks,
    color="skyblue",       # Fill color of the points
    s=100,                 # Size of the points
    alpha=0.7,             # Transparency
    edgecolor="black",     # Border color of the points
    linewidths=2           # controls the border width of the points
)

plt.show()
````

### **Title, Axis Labels, and Grid**

````python
import matplotlib.pyplot as plt

hours_studied = [1, 2, 3, 4, 5, 6, 7]
marks = [35, 42, 48, 55, 63, 71, 78]

plt.scatter(
    hours_studied,
    marks,
    color="skyblue",
    s=100,
    alpha=0.7,
    edgecolor="black",
    linewidths=2
)

plt.title("Study Hours vs Exam Marks")   # Gives the entire plot a title
plt.xlabel("Hours Studied")              # Labels the x-axis
plt.ylabel("Exam Marks")                 # Labels the y-axis

plt.grid()                               # Adds grid lines to the plot

plt.show()
````

### **c in scatter plot**
```text
color points based on data
Unlike color="blue", c=marks lets each point get a color based on its value.

````python
import matplotlib.pyplot as plt

hours_studied = [1, 2, 3, 4, 5, 6, 7]
marks = [35, 42, 48, 55, 63, 71, 78]

# c=marks → each point's color is based on its marks value
plt.scatter(
    hours_studied,
    marks,
    c=marks,
    s=100
)

plt.title("Study Hours vs Exam Marks")
plt.xlabel("Hours Studied")
plt.ylabel("Exam Marks")

plt.show()
````

### **cmap**

### **colorbar()** 
```text
It shows what the colors mean
When we use c=marks, the points get different colors, but someone looking at the graph doesn't know which color represents which mark.
```
`plt.colorbar()` adds that information.

````python
import matplotlib.pyplot as plt

hours_studied = [1, 2, 3, 4, 5, 6, 7]
marks = [35, 42, 48, 55, 63, 71, 78]

plt.scatter(
    hours_studied,
    marks,
    c=marks,          # Color each point according to its marks
    cmap="viridis",   # Choose the color gradient
    s=100
)

plt.colorbar(label="Marks")  # Shows what each color represents

plt.title("Study Hours vs Exam Marks")
plt.xlabel("Hours Studied")
plt.ylabel("Exam Marks")

plt.show()
````

### **vmin and vmax**
```text
vmin and vmax control which data values correspond to the bottom and top of the colormap.

vmin=30                         vmax=80
   ↓                               ↓
lowest color  ────────────────  highest color

````python
import matplotlib.pyplot as plt

hours_studied = [1, 2, 3, 4, 5, 6, 7]
marks = [35, 42, 48, 55, 63, 71, 78]

plt.scatter(
    hours_studied,
    marks,
    c=marks,           # Use marks to determine colors
    cmap="viridis",    # Choose the color gradient
    vmin=30,           # Treat 30 as the lowest color value
    vmax=80,           # Treat 80 as the highest color value
    s=100
)

plt.colorbar(label="Marks")

plt.title("Study Hours vs Exam Marks")
plt.xlabel("Hours Studied")
plt.ylabel("Exam Marks")

plt.show()
````

### **Multiple groups in one scatter plot**

````python
import matplotlib.pyplot as plt

# Study hours for Class A students
hours_a = [1, 2, 3, 4, 5]

# Marks for Class A students
marks_a = [35, 42, 48, 55, 63]

# Study hours for Class B students
hours_b = [2, 3, 4, 5, 6]

# Marks for Class B students
marks_b = [40, 50, 58, 68, 75]

# Plot Class A
plt.scatter(
    hours_a,
    marks_a,
    color="blue",
    s=100,
    label="Class A"
)

# Plot Class B
plt.scatter(
    hours_b,
    marks_b,
    color="orange",
    s=100,
    label="Class B"
)

plt.title("Study Hours vs Exam Marks")
plt.xlabel("Hours Studied")
plt.ylabel("Exam Marks")

plt.legend()  # Shows which color represents each class
plt.grid(axis="y", alpha=0.5)

plt.show()
````

**Interesting way of Multiple Scatter Plot**
```text
we can use one single scatter plot and define both the scatter plot in them with the help of class code.

Here:

class_code = [0, 0, 0, 1, 1, 1]

means:

0 → Class A
1 → Class B

````python
import matplotlib.pyplot as plt

hours = [1, 2, 3, 4, 5, 6]
marks = [35, 42, 50, 58, 65, 72]

# 0 = Class A
# 1 = Class B
class_code = [0, 0, 0, 1, 1, 1]

plt.scatter(
    hours,
    marks,
    c=class_code,       # Color is based on the category code
    cmap="viridis",     # Convert the codes into colors
    s=100
)

plt.title("Study Hours vs Exam Marks")
plt.xlabel("Hours Studied")
plt.ylabel("Exam Marks")

plt.show()
````

```text
Important: this works, but for a beginner, using two scatter() calls with label + legend() is usually clearer when you have categorical groups.

### **annotate()** 
```text
It is used to add text pointing to a specific data point

````python
import matplotlib.pyplot as plt 
 
hours_studied = [1, 2, 3, 4, 5, 6, 7] 
marks = [35, 42, 48, 55, 63, 71, 78] 
 
plt.scatter( 
    hours_studied, 
    marks, 
    color="blue", 
    s=100 
) 
 
plt.annotate( 
    "Highest Marks",  # Text that will be displayed
    xy=(7, 78)        # (x, y) coordinates of the point to label
) 
 
plt.title("Study Hours vs Exam Marks") 
plt.xlabel("Hours Studied") 
plt.ylabel("Exam Marks") 
plt.grid(axis="y", alpha=0.5) 
 
plt.show()
````

```text
xytext — moving the annotation text

xy tells Matplotlib which point to annotate.

xytext tells Matplotlib where to place the text.

````python
import matplotlib.pyplot as plt

hours_studied = [1, 2, 3, 4, 5, 6, 7]
marks = [35, 42, 48, 55, 63, 71, 78]

plt.scatter(
    hours_studied,
    marks,
    color="blue",
    s=100
)

plt.annotate(
    "Highest Marks",
    xy=(7, 78),          # The actual point we want to label
    xytext=(5.8, 75)       # Where the annotation text should be placed
)

plt.title("Study Hours vs Exam Marks")
plt.xlabel("Hours Studied")
plt.ylabel("Exam Marks")
plt.grid(axis="y", alpha=0.5)

plt.show()
````

**arrowprops** --> connects the text to the point

````python
import matplotlib.pyplot as plt

hours_studied = [1, 2, 3, 4, 5, 6, 7]
marks = [35, 42, 48, 55, 63, 71, 78]

plt.scatter(
    hours_studied,
    marks,
    color="blue",
    s=100
)

plt.annotate(
    "Highest Marks",

    xy=(7, 78),           # The actual point being labeled
    xytext=(5, 75),       # Where the text will be placed

    # arrowprops controls the arrow connecting the text to the point
    arrowprops={
        "arrowstyle": "->"  # "->" creates an arrow pointing to the point
    }
)

plt.title("Study Hours vs Exam Marks")
plt.xlabel("Hours Studied")
plt.ylabel("Exam Marks")
plt.grid(axis="y", alpha=0.5)

plt.show()
````

**Customizing arrowprops**

````python
import matplotlib.pyplot as plt

hours_studied = [1, 2, 3, 4, 5, 6, 7]
marks = [35, 42, 48, 55, 63, 71, 78]

plt.scatter(
    hours_studied,
    marks,
    color="blue",
    s=100
)

plt.annotate(
    "Highest Marks",

    xy=(7, 78),          # Point being labeled
    xytext=(5, 75),      # Position of the annotation text

    arrowprops={
        "arrowstyle": "->",  # Arrow pointing toward the data point
        "color": "red",      # Color of the arrow
        "linewidth": 2       # Thickness of the arrow
    }
)

plt.title("Study Hours vs Exam Marks")
plt.xlabel("Hours Studied")
plt.ylabel("Exam Marks")
plt.grid(axis="y", alpha=0.5)

plt.show()
````

### **Trend Line**

````python
import matplotlib.pyplot as plt
import numpy as np

hours_studied = [1, 2, 3, 4, 5, 6, 7]
marks = [35, 42, 48, 55, 63, 71, 78]

plt.scatter(
    hours_studied,
    marks,
    color="blue",
    s=100
)

# Calculate a straight line that represents the overall trend y=mx+b here 1 tells Numpy to fit a straight line
m, b = np.polyfit(hours_studied, marks, 1)

# Draw the trend line
plt.plot(hours_studied, m * np.array(hours_studied) + b) 

# m * np.array(hours_studied) + b calculates the y values of that trend line.

plt.title("Study Hours vs Exam Marks")
plt.xlabel("Hours Studied")
plt.ylabel("Exam Marks")
plt.grid(axis="y", alpha=0.5)

plt.show()
````
