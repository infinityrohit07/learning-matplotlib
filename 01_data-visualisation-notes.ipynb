# Data Visualization

## What is Data Visualization?

**Data Visualization** is the process of representing data graphically so that patterns, relationships, trends, comparisons, and distributions can be understood more easily.

Instead of looking at raw numbers:

```text
January   → 100
February  → 120
March     → 90
April     → 150
May       → 180

we can represent the same data visually using a chart.

Sales
  |
180|                         ●
160|                    ●
140|               ●
120|       ●
100|  ●
 80|          ●
  |
  +--------------------------------
     Jan   Feb   Mar   Apr   May

## Why Do We Need Data Visualization?

Large datasets can be difficult to understand when viewed only as tables or raw numbers.

**Data visualization helps us identify:**

- **Trends** — How something changes over time
- **Patterns** — Repeated or noticeable behavior in data
- **Comparisons** — Differences between categories
- **Relationships** — How two variables may be related
- **Distributions** — How values are spread across a range
- **Outliers** — Values that are unusually high or low
- **Composition** — How a total is divided among categories

### Example

Suppose we have:

| Month | Sales |
|---|---:|
| January | 100 |
| February | 120 |
| March | 90 |
| April | 150 |
| May | 180 |

A table tells us the **exact values**.

A chart can make the **increase, decrease, and overall trend** much easier to see.

This is why visualization is useful: it allows us to understand the **structure and behavior of data more quickly** than by examining individual values alone.

## Understanding Data Before Visualizing

Before creating any visualization, we need to understand the **data we are working with**.

Choosing a visualization depends on:

1. The type of data we have
2. The relationship between the variables
3. The question we want to answer       



A good visualization starts with:

* Understand the Data
* Identify Data Types
* Understand the Relationship
* Define the Question
* Choose the Visualization

## Type of Data

There are two broad categories:

1. Categorical Data
2. Numerical Data

### 1. *Categorical Data*

Categorical data represents groups, labels, or categories.       
These values describe what something is rather than representing a measurable numerical quantity.

Examples:     
     
City     
Product Category     
Gender      
Department     
Payment Method       
Color     
Country

### 2. *Numerical Data*

Numerical data represents quantities that can be counted or measured.

Examples:            

Age        
Height      
Weight       
Temperature      
Price             
Salary                   
Number of Products Sold     
Distance              

Numerical data is divided into two important types:           
* Discrete Data
* Continuous Data

#### 2.1 Discrete Numerical Data

Discrete data consists of values that are countable.    
These values usually represent the number of something.

Examples:    
Number of students  
Number of products sold     
Number of employees     
Number of orders        
Number of cars  
Number of rooms 


For example:    
Products Sold = 25      

Possible values could be:           
24,25,26 or 27

Normally, we would not say:     
25.73 products          
because products are counted as whole units.            

Therefore:              
Discrete data = Countable values            

#### 2.2 Continuous Numerical Data

Continuous data represents values that can be measured and can take many possible values within a range.

Examples:   
Height      
Weight      
Temperature         
Distance        
Time        
Speed       

For example, a person's weight could be:        
55 kg           
55.2 kg         
55.25 kg            
55.253 kg           
The value can have decimal places depending on the precision of the measurement.        

Therefore:          
Continuous data = Measurable values

## Example Dataset

Consider the following dataset:

| City | Product Category | Products Sold | Weight | Temperature |
|---|---|---:|---:|---:|
| Delhi | Electronics | 120 | 2.5 | 32.4 |
| Mumbai | Clothing | 85 | 0.8 | 29.7 |
| Pune | Electronics | 150 | 3.2 | 27.8 |
| Jaipur | Furniture | 60 | 12.5 | 35.1 |

This dataset contains **5 variables**, and each variable has a different type of data.

## Classifying the Variables

| Variable | Data Type | Reason |
|---|---|---|
| City | Categorical | Represents locations |
| Product Category | Categorical | Represents groups |
| Products Sold | Numerical - Discrete | Represents a count |
| Weight | Numerical - Continuous | Represents a measurement |
| Temperature | Numerical - Continuous | Represents a measurement |

### Why?

- **City** contains names of locations, so it is **categorical data**.
- **Product Category** contains groups such as Electronics, Clothing, and Furniture, so it is **categorical data**.
- **Products Sold** represents a count of products, so it is **discrete numerical data**.
- **Weight** represents a measurable quantity and can have decimal values, so it is **continuous numerical data**.
- **Temperature** is also a measurable quantity that can have decimal values, so it is **continuous numerical data**.

This classification helps us understand what kind of relationships and visualizations can be used with the variables.

## Data Visualisation Types

## 1. *Bar Chart*

A **bar chart** is a visualization used to compare a numerical value across different **categories**.

Each category is represented by a bar, and the **height or length of the bar represents the numerical value** associated with that category.

##### Data Relationship

A bar chart is commonly used for:

```text
Categorical Variable → Numerical Variable


What Can a Bar Chart Show?    
A bar chart can help us identify:
* Comparisons — Differences between categories
* Ranking — Which categories have higher or lower values
* Maximum and Minimum — The categories with the largest or smallest values
* Magnitude — The numerical size associated with each category
* Differences — How much values differ between categories

## 2. *Line Chart*

A **line chart** is a visualization used to show how a numerical value changes across an **ordered sequence**, especially over time.

A line chart represents individual data points and connects them with lines. The connected points make it easier to identify **trends, increases, decreases, and changes**.

##### Data Relationship

A line chart is commonly used for:

```text
Ordered Variable → Numerical Variable

What Can a Line Chart Show?
A line chart can help us identify:
* Trends — The general direction of change
* Increases — Values moving upward
* Decreases — Values moving downward
* Changes over time — How a value evolves across time
* Patterns — Repeated or noticeable movements
* Peaks and Troughs — High and low points in a sequence
* Rate of Change — How quickly or slowly values change

## 3. *Scatter Plot*

A **scatter plot** is a visualization used to examine the **relationship between two numerical variables**.

Each observation is represented by a **point** on the graph.

The position of each point is determined by the values of two numerical variables:

- **X-axis** → First numerical variable
- **Y-axis** → Second numerical variable
- **Each point** → One observation

### Data Relationship

A scatter plot is commonly used for:

```text
Numerical Variable → Numerical Variable

What Can a Scatter Plot Show?
A scatter plot can help us identify:
* Relationship — Whether two variables appear to be related
* Correlation — Whether variables tend to increase or decrease together
* Direction — Whether the relationship appears positive or negative
* Clusters — Groups of observations with similar characteristics
* Outliers — Observations that are far away from the general pattern
* Patterns — General structures or trends in the observations

## 4. *Pie Chart*

A **pie chart** is a visualization used to show how a **whole is divided into different categories**.

The entire circle represents **100% of the data**, and each slice represents the **proportion or percentage** of a category.

### Data Relationship

A pie chart is commonly used for:

```text
Categorical Variable → Numerical Value
                         ↓
                  Part of a Whole

What Can a Pie Chart Show?
A pie chart can help us identify:
* Proportion — How much each category contributes to the whole
* Percentage — The percentage represented by each category
* Composition — How a whole is divided among different categories
* Largest Share — Which category contributes the largest portion
* Smallest Share — Which category contributes the smallest portion
* Relative Contribution — How large one category is compared with the whole


## 5. *Histogram*

A **histogram** is a visualization used to show the **distribution of a numerical variable**.

It groups numerical values into **intervals called bins** and uses bars to show the frequency of observations within each interval.

### Data Relationship

A histogram is commonly used for:

```text
Numerical Variable
        ↓
      Bins
        ↓
    Frequency

What Can a Histogram Show?
A histogram can help us identify:
* Distribution — How numerical values are spread
* Frequency — How many observations fall within each range
* Center — Where most values are concentrated
* Spread — How widely the values are distributed
* Shape — The overall shape of the distribution
* Skewness — Whether the distribution is pulled toward one side
* Peaks — Ranges where many observations are concentrated
* Gaps — Ranges where few or no observations occur
* Possible Outliers — Values that appear far away from the main distribution


## 6. *Box Plot*

A **box plot** is a visualization used to summarize the **distribution of numerical data** using its median, quartiles, spread, and potential outliers.

### Data Relationship

A box plot is commonly used for:

```text
Numerical Variable
        ↓
Statistical Summary
        ↓
     Box Plot


What Can a Box Plot Show?
A box plot can help us identify:
* Median — The middle value of the data
* Quartiles — How the data is divided into four parts
* Spread — How widely the values are distributed
* Interquartile Range (IQR) — The spread of the middle 50% of the data
* Potential Outliers — Values that are unusually far from the main distribution
* Skewness — Whether the distribution appears more spread toward one side
* Variation — How much the numerical values differ


## 7. *Area Chart*

An **area chart** is a visualization used to show how a numerical value **changes across an ordered sequence**, while emphasizing the magnitude of the values.

It is similar to a line chart, but the area below the line is filled.

### Data Relationship

An area chart is commonly used for:

```text
Ordered Variable / Time → Numerical Variable


What Can an Area Chart Show?
An area chart can help us identify:
* Trends — The general direction of change
* Growth — How values increase over time
* Decline — How values decrease over time
* Magnitude — The size of numerical values
* Changes over Time — How values evolve across an ordered sequence
* Cumulative Patterns — How quantities build up over time
* Composition — How multiple categories contribute to a total when using a stacked area chart
* Relative Contribution — How different categories contribute to the overall quantity

# Quick Visualization Guide

| Visualization | Main Purpose | Typical Data |
|---|---|---|
| Bar Chart | Compare categories | Categorical + Numerical |
| Line Chart | Show trends | Ordered/Time + Numerical |
| Scatter Plot | Show relationships | Numerical + Numerical |
| Pie Chart | Show composition | Categorical + Proportion |
| Histogram | Show distribution | Numerical |
| Box Plot | Show distribution and spread | Numerical |
| Area Chart | Show trends and magnitude | Ordered/Time + Numerical |



# Choosing a Visualization Based on the Question

Instead of memorizing chart names, think about the **question you want to answer**.

| Question | Suitable Visualization |
|---|---|
| Which category has the highest value? | Bar Chart |
| How does a value change over time? | Line Chart |
| Are two numerical variables related? | Scatter Plot |
| How is a total divided among categories? | Pie Chart |
| How are numerical values distributed? | Histogram |
| What is the spread and median of the data? | Box Plot |
| How does the magnitude change over time? | Area Chart |
