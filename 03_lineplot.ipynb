## 1. **Line Plot**

Basic syntax:       
```python
plt.plot(x, y)
```
Where:      
x → values placed on the X-axis         
y → values placed on the Y-axis

````python
import matplotlib.pyplot as plt
````

````python
months = ["Jan", "Feb", "Mar", "Apr"]
sales = [100, 150, 120, 180]

plt.plot(months, sales)

plt.show()
````

Important syntax to remember
```python
plt.plot(x, y)
```
The first argument is generally the X data, and the second is the Y data.

### **What happens if we pass only one value?**

````python
y = [10, 20, 15, 25]

plt.plot(y)
plt.show()
````

matplotlib automatically creates x-values on its own.

### **Note: When you provide both x and y, they normally need to have the same number of values or else it might throw error**

You don't trust me run the below code to check.

````python
months = ["Jan", "Feb", "Mar"]  # no. of elements in months (x-axis) = 3
sales = [100, 150, 120, 180]    # no. of elements in sales (y-axis) = 4 

plt.plot(months, sales)  # x!=y

plt.show() 
````

### **Markers**

By default:
```python
plt.plot(x, y)
```
draws a line connecting the points, but it doesn't clearly show each individual data point.
                

A marker puts a symbol at each data point.

Basic syntax
```python
plt.plot(x, y, marker="o")
```

````python
x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

plt.plot(x, y, marker="o")

plt.show()
````

### **Line Color**

Basic syntax: 
```python
plt.plot(x, y, color="red")
```

color tells Matplotlib what color the line should be.

````python
x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

plt.plot(x, y, marker="o", color="red")

plt.show()
````

### **Line Style**

Matplotlib uses the linestyle parameter to control the style of the line.

Basic Syntax:
```python
plt.plot(x, y, linestyle="--")

````python
x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

plt.plot(x, y, marker="o", color="green", linestyle="--") # dashed line

plt.show()
````

````python
x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

plt.plot(x, y, marker="o", color="green", linestyle="-") # solid line - the default behavior even if we don't specify it

plt.show()
````

````python
x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

plt.plot(x, y, marker="o", color="green", linestyle=":") # dotted line

plt.show()
````

````python
x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

plt.plot(x, y, marker="o", color="green", linestyle="-.") # Dasg-dot line

plt.show()
````

### **Line Width**

This helps to control the thickness of the line.

Basic Syntax:
```python
plt.plot(x, y, linewidth=3)

````python
x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

plt.plot(x, y, marker="o", color="brown", linestyle="-", linewidth=3)

plt.show()
````

````python
x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

plt.plot(x, y, marker="o", color="brown", linestyle="--", lw=2) # lw can be used as shorter way of linewidth

plt.show()
````

### **Marker Size**

It controls the size of the marker.

Basic Syntax:
```python
plt.plot(x, y, marker="o", markersize=10)

````python
x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

plt.plot(x, y, marker="o", color="blue", linestyle="--", linewidth=2, markersize=10)

plt.show()
````

````python
x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

plt.plot(x, y, marker="o", color="blue", linestyle="--", lw=2, ms=6) # ms is used as shorter way of markersize

plt.show()
````

### **Marker Color**

It helps to change the marker color so that it stays different than our line color.

Basic Syntax:
```python
plt.plot(x, y, color="blue", marker="o", markerfacecolor="red")

````python
x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

plt.plot(x, y, marker="o", color="blue", linestyle="-", linewidth=5, markersize=6, markerfacecolor="red")

plt.show()
````

### **Marker Edge Color**

It helps to change color of the marker border.

````python
x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

plt.plot(x, y, marker="o", color="blue", linestyle="-", linewidth=5, markersize=6, markerfacecolor="red", markeredgecolor="black")

# markeredgecolor helps to change the color of the outer line of our marker (the border)

plt.show()
````

````python
x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

plt.plot(x, y, marker="o", color="blue", linestyle="-", linewidth=5, markersize=6, mfc="red", mec="black")

# mfc shorter for markerfacecolor
# mec shorter for markeredgecolor

plt.show()
````

### **Marker Edge Width**

It helps to change the marker's border thickness.

````python
x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

plt.plot(x, y, marker="o", color="blue", linestyle="-", linewidth=5, markersize=6, markerfacecolor="red", markeredgecolor="black", markeredgewidth=3)

plt.show()
````

````python
x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

plt.plot(x, y, marker="o", color="blue", linestyle="-", linewidth=5, markersize=6, mfc="red", mec="black", mew=3)

# mfc shorter for markerfacecolor
# mec shorter for markeredgecolor
# mew shorter for markeredgewith

plt.show()
````

### **Format Strings**

Matplotlib provides a short way to specify the line color, marker, and line style together.

Basic Syntax:
```python
plt.plot(x, y, "format")

````python
x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

plt.plot(x, y, "ro-")  # ro-  --> r stands for red, o stands for marker type and - stands for solid line

plt.show()
````

````python
x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

plt.plot(x, y, "b*:")  # b*:  --> b stands for blue, * stands for marker type and : stands for dotted line

plt.show()
````

### **Title and Axis Labels**

````python
months = ["Jan", "Feb", "Mar", "Apr"]
sales = [100, 150, 120, 180]

plt.plot(months, sales, marker="o")

plt.title("Monthly Sales") # gives title for which our line plot stands for.
plt.xlabel("Month") # gives title to our x-axis for which it stands.
plt.ylabel("Sales") # gives title to our y-axis for which it stands.

plt.show()
````

### **Grid**

A grid adds horizontal and/or vertical lines to the plot. This makes it easier to read the values of individual data points.

````python
months = ["Jan", "Feb", "Mar", "Apr"]
sales = [100, 150, 120, 180]

plt.plot(months, sales, marker="o")

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.grid()

plt.show()
````

````python
months = ["Jan", "Feb", "Mar", "Apr"]
sales = [100, 150, 120, 180]

plt.plot(months, sales, marker="o")

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.grid(axis="x") # axis="x" gives horizontal grid

plt.show()
````

````python
months = ["Jan", "Feb", "Mar", "Apr"]
sales = [100, 150, 120, 180]

plt.plot(months, sales, marker="o")

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.grid(axis="y") # axis="y" gives vertical grid

plt.show()
````

````python
months = ["Jan", "Feb", "Mar", "Apr"]
sales = [100, 150, 120, 180]

plt.plot(months, sales, marker="o")

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.grid(axis="both") # the default behavior with both grid

plt.show()
````

### **Plotting Multiple Lines**

Its easy to plot multiple lines. Look at the example below and try to understand with every knowledge we have gained till now.

````python
months = ["Jan", "Feb", "Mar", "Apr"]

product_a = [100, 150, 120, 180]
product_b = [80, 130, 160, 140]

plt.plot(
    months,
    product_a,
    color="blue",
    marker="o"
)

plt.plot(
    months,
    product_b,
    color="red",
    marker="o"
)

plt.title("Monthly Product Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.grid()

plt.show()
````

We have our line plot but how will we know which line is product A and which line is product B. This is done by giving label to each of our line plot and then calling legend.         
Note: **A legend** identifies what each line, marker, or dataset represents in the plot.

````python
months = ["Jan", "Feb", "Mar", "Apr"]
product_a = [100, 150, 120, 180]
product_b = [80, 130, 160, 140]

plt.plot(
    months,
    product_a,
    color="blue",
    marker="o",
    label="Product A"
)

plt.plot(
    months,
    product_b,
    color="red",
    marker="o",
    label="Product B"
)

plt.title("Monthly Product Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.grid()
plt.legend()

plt.show()
````
