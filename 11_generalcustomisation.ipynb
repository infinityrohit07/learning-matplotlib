### **xlim() and ylim()**

These functions let you control the visible range of the X-axis and Y-axis.

`xlim()`

Controls the X-axis range:
`plt.xlim(min, max)`

`ylim()`

Controls the Y-axis range:
`plt.ylim(min, max)`

````python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

plt.plot(x, y)

plt.xlim(1, 5)
plt.ylim(0, 35)

plt.title("Sales")
plt.xlabel("Day")
plt.ylabel("Sales")

plt.show()
````

```text
so now x-axis starts from 1 and ends at 5.
and our y-axis starts from 0 and ends at 35. 
both were set using xlim() and ylim()

### **xticks() and yticks()**

`xticks()` and `yticks()` control the tick marks and their labels shown on that axis.

**Basic syntax:**
```python
plt.xticks(values)
plt.yticks(values)

````python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [100, 120, 90, 150, 180]

plt.plot(months, sales)

#plt.xticks(months) # this acts as default as we pass the same value again.
plt.xticks(
    months,
    [0, 1, 2, 3, 4] # this takes months and replaces it with 0, 1, 2, 3, 4
)
plt.yticks([0, 50, 100, 150, 200])

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
````

```text
with xticks() and yticks() we can set what should exactly be visible on our x-axis and y-axis (the ticks and their labels)

### **Legend customization**
```text
A legend explains which line, bar, or data series represents what.
For a legend to work we must always specify the labels for each line, bar or data series.

````python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]

sales = [100, 120, 90, 150, 180]
profit = [20, 30, 15, 40, 50]

plt.plot(months, sales, label="Sales")
plt.plot(months, profit, label="Profit")

plt.legend()

plt.show()
````

**Legend positioning and title**
```text
plt.legend(loc="upper left", title="title_we_want")

Common locations:
"upper right"
"upper left"
"lower right"
"lower left"
"center"

````python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]

sales = [100, 120, 90, 150, 180]
profit = [20, 30, 15, 40, 50]

plt.plot(months, sales, label="Sales")
plt.plot(months, profit, label="Profit")

plt.legend(loc="center", title="Categories")

plt.show()
````
