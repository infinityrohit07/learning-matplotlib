## **Matplotlib + Pandas**

**Pandas** and **Matplotlib** can be used together to make data visualization easier.

Pandas provides the **DataFrame** structure for storing and manipulating data, while Matplotlib provides the tools for creating visualizations. Pandas has built-in plotting methods such as `df.plot()` that use Matplotlib in the background.

This allows us to create common charts directly from a DataFrame without manually passing each list of values to Matplotlib.

Pandas plotting is useful when:

- Your data is already stored in a **DataFrame**.
- You want to quickly visualize numerical columns.
- You want to plot data after **filtering, grouping, or transforming** it with Pandas.
- You want to create common plots such as **line charts, bar charts, histograms, and scatter plots**.

### Basic idea

```text
Pandas DataFrame
       ↓
   Data cleaning
       ↓
 Filtering / Grouping
       ↓
     df.plot()
       ↓
    Matplotlib
       ↓
    Visualization

### **df.plot() — Basic Syntax**

````python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "month": ["Jan", "Feb", "Mar", "Apr", "May"],
    "sales": [100, 120, 90, 150, 180]
})

df.plot()

plt.show()
````

### **Choosing the plot type with kind**

By default, `df.plot()` creates a line plot.

You can choose another plot type using the kind parameter:
```python
df.plot(
    x="month",
    y="sales",
    kind="bar"
)
```

| `kind`      | Plot         |
| ----------- | ------------ |
| `"line"`    | Line plot    |
| `"bar"`     | Bar chart    |
| `"hist"`    | Histogram    |
| `"scatter"` | Scatter plot |


````python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "month": ["Jan", "Feb", "Mar", "Apr", "May"],
    "sales": [100, 120, 90, 150, 180]
})

# Line plot
df.plot(
    x="month",
    y="sales",
    kind="line"
)

plt.title("Monthly Sales - Line Plot")
plt.show()


# Bar plot
df.plot(
    x="month",
    y="sales",
    kind="bar"
)

plt.title("Monthly Sales - Bar Plot")
plt.show()
````

### **Scatter Plot with Panda**

````python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "hours_studied": [1, 2, 3, 4, 5, 6, 7],
    "marks": [35, 42, 48, 55, 63, 71, 78]
})

df.plot(
    x="hours_studied",
    y="marks",
    kind="scatter",
    s=100,
    alpha=0.7,
    color="steelblue",
    edgecolor="black"
)

plt.title("Study Hours vs Marks")
plt.xlabel("Hours Studied")
plt.ylabel("Marks")

plt.show()
````

### **Multiple columns with df.plot()**

````python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "month": ["Jan", "Feb", "Mar", "Apr", "May"],
    "sales": [100, 120, 90, 150, 180],
    "profit": [20, 30, 15, 40, 50]
})

df.plot(
    x="month",
    y=["sales", "profit"]
)

plt.title("Sales and Profit")
plt.xlabel("Month")
plt.ylabel("Amount")

plt.show()
````

### **Basic customization with Pandas plots**
```text
We have already learned these in Matplotlib, so here we're just learning how they work with df.plot().

````python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "month": ["Jan", "Feb", "Mar", "Apr", "May"],
    "sales": [100, 120, 90, 150, 180]
})

df.plot(
    x="month",
    y="sales",
    kind="line",
    figsize=(8, 5),
    color="green",
    marker="o",
    linewidth=2,
    grid=True
)

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
````

### **groupby() + plotting**

````python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "product": ["Laptop", "Phone", "Laptop", "Phone", "Tablet", "Tablet"],
    "month": ["Jan", "Jan", "Feb", "Feb", "Jan", "Feb"],
    "sales": [100, 150, 120, 170, 80, 100]
})

print(df)
````

````python
product_sales = df.groupby("product")["sales"].sum()

print(product_sales)
````

````python
product_sales.plot(
    kind="bar",
    figsize=(7, 5)
)

plt.title("Total Sales by Product")
plt.xlabel("Product")
plt.ylabel("Sales")

plt.savefig("total_sales.png")

plt.show()
````
