### 7. **Area / Filled Plots**

An **area plot** is a line plot where the area under the line is filled with color.
```text
It is useful for showing:

Trends over time
Magnitude of values
Cumulative changes
How much of something exists over a range
```
In Matplotlib, we commonly create basic filled areas using `plt.fill_between()`.


**Basic syntax:**
```python
plt.fill_between(x, y)
```
```text
Where:

x → values for the x-axis
y → values for the y-axis
The area between y and 0 is filled.

````python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 15, 12, 20, 18]

plt.fill_between(x, y)

plt.show()
````

### **Color**

`plt.fill_between(x, y, color="color_name")`

````python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 15, 12, 20, 18]

plt.fill_between(x, y, color="skyblue")

plt.title("Sales Over 5 Days")
plt.xlabel("Day")
plt.ylabel("Sales")

plt.show()
````

### **alpha**
```text
alpha values
alpha=1 → completely opaque
alpha=0.5 → 50% transparent
alpha=0 → completely transparent

````python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 15, 12, 20, 18]

plt.fill_between(
    x,
    y,
    color="skyblue",
    alpha=0.4
)

plt.title("Sales Over 5 Days")
plt.xlabel("Day")
plt.ylabel("Sales")

plt.show()
````

### **y2 — Fill between two values**
```text
So far:

plt.fill_between(x, y)

fills the area between y and 0.

But we can also fill the area between two lines using y2.
```
Syntax:
`plt.fill_between(x, y1, y2)`
```text
x → x-axis values
y1 → first boundary
y2 → second boundary

````python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]

y1 = [10, 15, 12, 20, 18]
y2 = [5, 8, 7, 10, 12]

plt.fill_between(x, y1, y2, color="skyblue", alpha=0.5)

plt.plot(x, y1)
plt.plot(x, y2)

plt.title("Area Between Two Lines")
plt.xlabel("Day")
plt.ylabel("Value")

plt.show()
````

### **where**
```text
where lets us fill only certain parts of the area based on a condition.

This is useful when we want to highlight a specific region of our data.
```
Basic syntax:
`plt.fill_between(x, y, where=condition)`

````python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 15, 12, 20, 18]

plt.plot(x, y)

plt.fill_between(
    x,
    y,
    where=[value > 15 for value in y],
    alpha=0.5
)

plt.title("Sales Above 15")
plt.xlabel("Day")
plt.ylabel("Sales")

plt.show()
````

### **interpolate**

The `interpolate` parameter is used with `plt.fill_between()` when a `where` condition is used to fill only a specific part of a plot. \
By default, Matplotlib determines the filled region using the existing data points. 
However, a line may cross the boundary defined by the condition between two data points. 
Setting `interpolate=True` makes Matplotlib calculate that crossing point and extend the filled area up to the actual intersection. 
This makes the boundary of the filled region more accurate.

```python
plt.fill_between(x, y, 15, where=condition, interpolate=True)

````python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 15, 12, 20, 18]

plt.plot(x, y)

plt.fill_between(
    x,
    y,
    15, #y2
    where=[value > 15 for value in y],
    interpolate=True,
    alpha=0.5
)

plt.axhline(15, linestyle="--")

plt.title("Values Above 15")
plt.xlabel("Day")
plt.ylabel("Value")

plt.show()
````

## **plt.stackplot()**
```text
A stacked area plot shows multiple datasets as filled areas stacked on top of each other. It is useful when you want to see how individual categories contribute to a combined total and how those contributions change over time.
```
**Basic syntax:**
`plt.stackplot(x, y1, y2, y3)`
```text
x → x-axis values
y1, y2, y3 → values for each category
Each dataset is stacked on top of the previous one.

````python
import matplotlib.pyplot as plt

months = [1, 2, 3, 4, 5]

product_a = [10, 15, 12, 18, 20]
product_b = [5, 8, 10, 12, 15]
product_c = [3, 5, 7, 8, 10]

plt.stackplot(
    months,
    product_a,
    product_b,
    product_c,
    labels=["Product A", "Product B", "Product C"]
)

plt.title("Product Sales Over Time")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.legend()
plt.show()
````

### **Color and alpha**

````python
import matplotlib.pyplot as plt

months = [1, 2, 3, 4, 5]

product_a = [10, 15, 12, 18, 20]
product_b = [5, 8, 10, 12, 15]
product_c = [3, 5, 7, 8, 10]

plt.stackplot(
    months,
    product_a,
    product_b,
    product_c,
    labels=["Product A", "Product B", "Product C"],
    colors=["skyblue", "lightgreen", "orange"],
    alpha=0.5
)

plt.title("Product Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.legend()

plt.show()
````

### **baseline**
```text
baseline controls where the stacked areas start on the y-axis.

The default is "zero", meaning the stack starts from 0.

For beginner Matplotlib, you only need to know the two useful options:

"zero" → starts from 0
"sym" → creates a symmetric baseline around zero

````python
import matplotlib.pyplot as plt

months = [1, 2, 3, 4, 5]

product_a = [10, 15, 12, 18, 20]
product_b = [5, 8, 10, 12, 15]
product_c = [3, 5, 7, 8, 10]

plt.stackplot(
    months,
    product_a,
    product_b,
    product_c,
    labels=["Product A", "Product B", "Product C"],
    baseline="sym",
    alpha=0.6
)

plt.title("Product Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.legend()

plt.show()
````
