## 6. **BOX PLOTS**
```text
A box plot is used to understand the distribution of numerical data.

It helps us see:

Median → middle value
Spread → how much the data varies
Quartiles → how the data is divided
Outliers → unusually high or low values
```
**Basic syntax:**
```python
plt.boxplot(data)

````python
import matplotlib.pyplot as plt

marks = [45, 50, 52, 55, 58, 60, 62, 65, 68, 90]

plt.boxplot(marks)

plt.show()
````

#### 1. **Understanding The Box**
```text
        ─────  ← upper cap
          |
          |    ← upper whisker
       ┌─────┐ <- Q3
       │     │
       │─────│  ← median
       │     │
       └─────┘ <- Q1
          |
          |    ← lower whisker
        ─────  ← lower cap
```

#### 2. **Data being divided by position:**
```text
Lowest 25%       Middle 50%        Highest 25%
    ↓                ↓                  ↓

45  50  |  52  55  58  60  62  65  |  68  90
         ↑                            ↑
        Q1                           Q3
```
```text
Q1 = boundary after the lowest 25%
Q3 = boundary after the lowest 75%
Box (Q1 → Q3) = middle 50% of the data points
The remaining data is represented by the whiskers/outlier.
```

#### 3. **Our Plot**

```text
90   ○       ← Outlier

68  ───      ← Upper cap
     |
     |
   ┌─────┐   ← Q3
   │     │
59 │─────│   ← Median
   │     │
   └─────┘   ← Q1
     |
     |
45  ───      ← Lower cap


The 90 is outside the whisker range, so Matplotlib displays it as a separate circle.

### **IQR (Interquartile Range)**
```text
The IQR tells us how spread out the middle 50% of the data is.

Formula:

IQR = Q3 − Q1

For our box plot:

Q1 → bottom of the box
Q3 → top of the box
IQR → height of the box
Why do we need IQR?

Matplotlib uses IQR to determine which values are potential outliers.

The common rule is:

Lower limit = Q1 − 1.5 × IQR
Upper limit = Q3 + 1.5 × IQR

Values outside these limits are shown as outliers.
In our case, 90 is outside the limits as a result it is shown as outliners.
Lets calculate the limits and see.

````python
import numpy as np

marks = [45, 50, 52, 55, 58, 60, 62, 65, 68, 90]

Q1 = np.percentile(marks, 25) # gives value exactly at 25% 
Q3 = np.percentile(marks, 75) # gives value exactly at 75%

IQR = Q3 - Q1 # subtracting both gives height of box

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
````

```text
now we calulate the limits using conventional rule introduced by statistician John Tukey as part of the box-and-whisker plot method.
The factor 1.5 is a practical cutoff chosen by Tukey.

````python
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Lower limit:", lower_limit)
print("Upper limit:", upper_limit)
````

```text
Values below 38.5 → potential outliers
Values above 78.5 → potential outliers
90 > 78.5 → 90 is an outlier

### **Multiple Box Plots for Comparisons**

````python
class_a = [45, 50, 52, 55, 58, 60, 62, 65, 68]
class_b = [55, 60, 62, 65, 67, 70, 72, 75, 80]

plt.boxplot([class_a, class_b])

plt.show()
````

### **Adding Labels to our Multiple Box Plots**

````python
import matplotlib.pyplot as plt

class_a = [45, 50, 52, 55, 58, 60, 62, 65, 68]
class_b = [55, 60, 62, 65, 67, 70, 72, 75, 80]

plt.boxplot(
    [class_a, class_b],
    tick_labels=["Class A", "Class B"]
)

plt.ylabel("Marks")
plt.xlabel("Classes")
plt.title("Marks Distribution")

plt.show()
````

### **Horizontal Box Plot**

````python
import matplotlib.pyplot as plt

marks = [45, 50, 52, 55, 58, 60, 62, 65, 68, 90]

plt.boxplot(
    marks,
    orientation="horizontal"
)

plt.xlabel("Marks")
plt.title("Marks Distribution")

plt.show()
````

### **Customize the box color**
```text
By default, Matplotlib gives the box its standard appearance. We can use patch_artist=True to allow the box itself to be filled.

plt.boxplot(
    marks,
    patch_artist=True
)

Then we can set the box color using boxprops:

plt.boxplot(
    marks,
    patch_artist=True,
    boxprops={"facecolor": "lightblue"}
)


patch_artist=True → allows the box to be filled
boxprops → controls the box appearance
facecolor → sets the box's fill color

````python
import matplotlib.pyplot as plt

marks = [45, 50, 52, 55, 58, 60, 62, 65, 68, 90]

plt.boxplot(
    marks,
    patch_artist=True,
    boxprops={"facecolor": "lightblue"}
)

plt.title("Marks Distribution")
plt.ylabel("Marks")

plt.show()
````

### **Customizing the Median Line**

````python
import matplotlib.pyplot as plt

marks = [45, 50, 52, 55, 58, 60, 62, 65, 68, 90]

plt.boxplot(
    marks,
    patch_artist=True,
    boxprops={"facecolor": "black", "alpha":0.2},
    medianprops={"color": "yellow", "linewidth": 2}
)

plt.title("Marks Distribution")
plt.ylabel("Marks")

plt.show()
````

### **Customizing the Whiskers**

````python
import matplotlib.pyplot as plt

marks = [45, 50, 52, 55, 58, 60, 62, 65, 68, 90]

plt.boxplot(
    marks,
    patch_artist=True,
    boxprops={"facecolor": "green", "alpha":0.3},
    medianprops={"color": "red", "linewidth": 2},
    whiskerprops={"color": "green", "linewidth": 3}
)

plt.title("Marks Distribution")
plt.ylabel("Marks")

plt.show()
````

### **Customizing the Caps**

````python
import matplotlib.pyplot as plt

marks = [45, 50, 52, 55, 58, 60, 62, 65, 68, 90]

plt.boxplot(
    marks,
    patch_artist=True,
    boxprops={"facecolor": "green", "alpha":0.3},
    medianprops={"color": "red", "linewidth": 2},
    whiskerprops={"color": "green", "linewidth": 3},
    capprops={"color": "black", "linewidth": 4}
)

plt.title("Marks Distribution")
plt.ylabel("Marks")

plt.show()
````

## **PRACTISE EXERCISE**

Create a box plot for these three classes:
```PYTHON
class_a = [45, 50, 52, 55, 58, 60, 62, 65, 68, 90]

class_b = [50, 54, 57, 60, 62, 65, 67, 70, 72, 75]

class_c = [40, 45, 48, 50, 53, 55, 57, 60, 62, 65]
```
```TEXT
Your plot should have:

All 3 classes in one box plot
Labels: Class A, Class B, Class C
Title: Marks Distribution
Y-axis label: Marks
Light-colored box fill
Red median line
Grid on the Y-axis

````python
import matplotlib.pyplot as plt

class_a = [45, 50, 52, 55, 58, 60, 62, 65, 68, 90]
class_b = [50, 54, 57, 60, 62, 65, 67, 70, 72, 75]
class_c = [40, 45, 48, 50, 53, 55, 57, 60, 62, 65]

plt.boxplot(
    [class_a,class_b,class_c],
    tick_labels=["Class A", "Class B", "Class C"],
    patch_artist=True,
    boxprops={"facecolor":"green", "alpha":0.3},
    medianprops={"color":"red", "linewidth":1, "linestyle":"--"}
)

plt.title("Marks Distribution")
plt.ylabel("Marks")
plt.grid(axis="y")

plt.show()
````
