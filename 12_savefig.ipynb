## **Saving figures with savefig()**

This lets us save our plot as an image/file instead of only displaying it in Jupyter.

**Basic syntax:**
```python
plt.savefig("sales_plot.png")

````python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [100, 120, 90, 150, 180]

plt.plot(months, sales)

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.savefig("sales_plot.png")

plt.show()
````

### **format**
```text
Common formats:
```
```python
plt.savefig("sales_plot.png")   # PNG
plt.savefig("sales_plot.jpg")   # JPEG
plt.savefig("sales_plot.pdf")   # PDF
plt.savefig("sales_plot.svg")   # SVG
```
```text
These directly tells Matplotlib to save our plots in which format.
But we can also explicitly specify them.
```
`plt.savefig("sales_plot", format="png")`


````python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [100, 120, 90, 150, 180]

plt.plot(months, sales)

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.savefig("sales_plot", format="pdf")

plt.show()
````

### **dpi**
```text
dpi means dots per inch. In savefig(), it controls the resolution/detail of the saved image.

Higher dpi generally means:
More pixels
Sharper image
Larger file size
```

|      DPI | Use                           |
| -------: | ----------------------------- |
| `80–100` | Basic/screen use              |
|    `150` | Good general quality          |
|    `300` | High-quality reports/printing |

````python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [100, 120, 90, 150, 180]

plt.plot(months, sales)

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.savefig("sales_plot_dpi.png", dpi=150)

plt.show()
````

### **bbox_inches="tight"**

Sometimes a saved figure has extra empty space around the plot, especially when you have long labels or titles.
```python
plt.savefig(
    "sales_plot.png",
    bbox_inches="tight"
)

````python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [100, 120, 90, 150, 180]

plt.plot(months, sales)

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.savefig("sales_plot_bbox.png", bbox_inches="tight")

plt.show()
````
