# Matplotlib

## 1. What is Matplotlib?

Matplotlib is a Python library used to create data visualizations.

It allows us to turn numerical and categorical data into visual forms such as:

* Line charts
* Bar charts
* Scatter plots
* Histograms
* Pie charts
* Box plots
* Area charts

## 2. Why do we use Matplotlib?

Matplotlib helps us:

* Understand data more easily
* Identify trends
* Compare values
* Find patterns
* Spot unusual values (outliers)
* Understand distributions
* Communicate results visually

## 3. Where does Matplotlib fit with NumPy and Pandas?

Numpy -> Numerical Data -> Panda -> Clean/Analyse -> Matplotlib -> Visualised Data

## 8. Understanding a matplotlib plot 

A Matplotlib visualization can be thought of like this:

```text
Figure
│
└── Axes
    │
    ├── X-axis
    ├── Y-axis
    ├── Title
    ├── Data
    ├── Legend
    └── Grid
```

### 1. Figure

A **Figure** is the overall container for your visualization.

Think of it as the **entire canvas/page**.

```text
┌──────────────────────────────────────────┐
│                  Figure                  │
│                                          │
│       ┌──────────────────────┐           │
│       │        Axes          │           │
│       │                      │           │
│       │      Plot Area       │           │
│       │                      │           │
│       └──────────────────────┘           │
│                                          │
└──────────────────────────────────────────┘
```

A Figure can contain **one or multiple Axes**.

### 2. Axes

This is where the actual graph is drawn.
This entire plotting area is an Axes.

Axes ≠ axis

* Axes → the actual plotting area
* Axis → X-axis or Y-axis


### 3. X-axis

The **X-axis** is normally horizontal.

```text
──────────────────────────────→
            X-axis
```

It commonly represents:

- Time
- Categories
- Independent variable

**Example:**

```text
Jan    Feb    Mar    Apr
```

### 4. Y-axis

The **Y-axis** is normally vertical.

```text
        ↑
        │
        │  Y-axis
        │
        │
        └────────────────→
```

It commonly represents:

- Sales
- Temperature
- Marks
- Quantity
- Another numerical measurement

### 5. Title

The **title** tells us **what the graph represents**.

**Example:**

```text
Monthly Sales
      ↑
    Title
```

**In Matplotlib:**

```python
plt.title("Monthly Sales")
```

### 6. Labels

We can label both axes.

```python
plt.xlabel("Month")
plt.ylabel("Sales")
```

This makes the graph **easier to understand**.

### 7. Legend

Suppose you have **multiple lines**:

```text
Sales
  ↑
  │      ─────  Product A
  │      ─────  Product B
  │
  └────────────────→
```

The **legend** tells us **which line represents what**.

**In Matplotlib:**

```python
plt.legend()
```

We'll learn this properly when we plot **multiple datasets**.

### 8. Grid

A **grid** helps us read values from the graph more easily.

**In Matplotlib:**

```python
plt.grid()
```

It adds lines such as:

```text
30 ─────────────────
20 ─────────────────
10 ─────────────────
   ─────────────────
```
