## 5. **Pie Chart**

A **pie chart** is used to show how different categories make up a **whole**.

**Basic syntax:**

```python
plt.pie(values)

````python
import matplotlib.pyplot as plt

expenses = [4000, 8000, 2000, 1000]

plt.pie(expenses)

plt.show()
````

### **labels**

**Labels** tell us which category each slice represents.
Each value should have a corresponding label in the same position.

**Basic Syntax:**
```python
plt.pie(
    values,
    labels=["Category 1", "Category 2", "Category 3"]
)

````python
import matplotlib.pyplot as plt

expenses = [4000, 8000, 2000, 1000]

plt.pie(
    expenses,
    labels=["Food", "Rent", "Transport", "Shopping"]
)

plt.show()
````

### **autopct**

autopct displays the percentage of each category inside the pie chart.

**Basic Syntax:**
```python
plt.pie(
    values,
    labels=labels,
    autopct="%1.1f%%"
)
```

%1.1f → show 1 decimal place
%% → display the % symbol

Example: 25.0%, 37.5%, etc.

````python
import matplotlib.pyplot as plt

expenses = [4000, 8000, 2000, 1000]

plt.pie(
    expenses,
    labels=["Food", "Rent", "Transport", "Shopping"],
    autopct="%1.1f%%"
)

plt.show()
````

if we need 2 decimal places we can do %1.2f

### **startangle**

startangle controls where the first slice of the pie chart starts.

**Basic Syntax:**
```python
plt.pie(
    values,
    labels=labels,
    autopct="%1.1f%%",
    startangle=90
)
```
```text
startangle=0 → starts from the right
startangle=90 → starts from the top
startangle=180 → starts from the left

We can use any angle from 0 to 360.

````python
import matplotlib.pyplot as plt

expenses = [4000, 8000, 2000, 1000]

plt.pie(
    expenses,
    labels=["Food", "Rent", "Transport", "Shopping"],
    autopct="%1.1f%%",
    startangle=180
)

plt.show()
````

### **colors**

colors lets you set a different color for each slice.

**Basic Syntax:**
```python
plt.pie(
    values,
    labels=labels,
    colors=["red", "blue", "green", "orange"]
)
```

The colors follow the same order as your values and labels.

````python
import matplotlib.pyplot as plt

expenses = [4000, 8000, 2000, 1000]

plt.pie(
    expenses,
    labels=["Food", "Rent", "Transport", "Shopping"],
    autopct="%1.1f%%",
    startangle=180,
    colors=["red", "blue", "green", "orange"]
)

plt.show()
````

or create a colors list and pass it to colors of plt.pie() 

````python
import matplotlib.pyplot as plt

expenses = [4000, 8000, 2000, 1000]

colors = ["red", "blue", "green", "orange"]

plt.pie(
    expenses,
    labels=["Food", "Rent", "Transport", "Shopping"],
    autopct="%1.1f%%",
    startangle=180,
    colors= colors
)

plt.show()
````

### **explode**

explode lets you separate one or more slices from the center of the pie chart.

**Basic Syntax:**
```python
plt.pie(
    values,
    labels=labels,
    explode=[0, 0.1, 0, 0]
)
```
```text
Here:

0 → slice stays in its normal position
0.1 → slice moves outward
The values in explode correspond to the slices in the same order.

````python
import matplotlib.pyplot as plt

expenses = [4000, 8000, 2000, 1000]

colors = ["red", "blue", "green", "orange"]

plt.pie(
    expenses,
    labels=["Food", "Rent", "Transport", "Shopping"],
    autopct="%1.1f%%",
    startangle=180,
    colors= colors,
    explode=[0,0,0.1,0]
)

plt.show()
````

### **shadow**

shadow=True adds a shadow behind the pie chart, giving it a slightly 3D-like appearance.

**Basic Synatx:**
```python
plt.pie(
    values,
    labels=labels,
    shadow=True
)

````python
import matplotlib.pyplot as plt

expenses = [4000, 8000, 2000, 1000]

colors = ["red", "blue", "green", "orange"]

plt.pie(
    expenses,
    labels=["Food", "Rent", "Transport", "Shopping"],
    autopct="%1.1f%%",
    startangle=180,
    colors= colors,
    explode=[0,0,0.1,0],
    shadow=True
)

plt.show()
````

### **wedgeprops**
```text
wedgeprops is used to customize the border/edge of each pie slice.
Let's look at its syntax through code.

````python
import matplotlib.pyplot as plt

expenses = [4000, 8000, 2000, 1000]

colors = ["red", "blue", "green", "orange"]

plt.pie(
    expenses,
    labels=["Food", "Rent", "Transport", "Shopping"],
    autopct="%1.1f%%",
    startangle=180,
    colors= colors,
    explode=[0,0,0.1,0],
    shadow=True,
    wedgeprops={
        "edgecolor": "white",
        "linewidth": 2
    }
)

plt.show()
````

### **axis("equal")**

This makes sure the pie chart stays circular instead of becoming stretched/oval.

````python
import matplotlib.pyplot as plt

expenses = [4000, 8000, 2000, 1000]

colors = ["red", "blue", "green", "orange"]

plt.pie(
    expenses,
    labels=["Food", "Rent", "Transport", "Shopping"],
    autopct="%1.1f%%",
    startangle=180,
    colors= colors,
    explode=[0,0,0.1,0],
    shadow=True,
    wedgeprops={
        "edgecolor": "white",
        "linewidth": 2
    }
)

plt.axis("equal")

plt.show()
````

### **Legend**
```text
A legend is useful when:

The labels are too long and clutter the pie.
We want the category names outside the chart.
We want more control over where the category names appear.

Note: we need labels for legend.

````python
import matplotlib.pyplot as plt

expenses = [4000, 8000, 2000, 1000]
labels = ["Food", "Rent", "Transport", "Shopping"]

plt.pie(
    expenses,
    autopct="%1.1f%%",
    startangle=90
)

# Add category names as a legend
plt.legend(labels)

# Keep the pie circular
plt.axis("equal")

plt.show()
````
