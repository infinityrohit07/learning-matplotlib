### **Subplots**

A subplot allows us to display multiple plots inside a single figure.
This is useful when we want to compare different visualizations or datasets side by side instead of creating separate figures.

**Basic syntax:**

The modern and recommended approach is:

`fig, ax = plt.subplots()`

For multiple plots:

`fig, ax = plt.subplots(rows, columns)`

### 1. **One subplot**

`fig, ax = plt.subplots()`
```text
It creates:

fig → the entire figure/window
ax → the plot area (axes) inside that figure
```
Then instead of using
`plt.plot()`

we can use:
`ax.plot()` 

````python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 15, 12, 20, 18]

fig, ax = plt.subplots()

ax.plot(x, y)

ax.set_title("Sales")
ax.set_xlabel("Day")
ax.set_ylabel("Sales")

plt.show()
````

```text
Notice the difference:

plt.title()
becomes:
ax.set_title()

and:
plt.xlabel()
plt.ylabel()
become:
ax.set_xlabel()
ax.set_ylabel()

### 2. **Multiple subplot**

`fig, ax = plt.subplots(rows, columns)`
```text
It creates:

fig → the entire figure/window
ax → the plot area (axes) inside that figure
row → number of rows
columns → number of columns
```

`plt.subplots(rows, columns)` determines how many subplot areas we create.

````python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]

sales = [10, 15, 12, 20, 18]
profit = [5, 8, 7, 12, 10]

fig, ax = plt.subplots(1, 2)

ax[0].plot(x, sales)
ax[0].set_title("Sales")

ax[1].plot(x, profit)
ax[1].set_title("Profit")

plt.show()
````

**Two rows and one column**

````python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]

sales = [10, 15, 12, 20, 18]
profit = [5, 8, 7, 12, 10]

fig, ax = plt.subplots(2, 1)

ax[0].plot(x, sales)
ax[0].set_title("Sales")

ax[1].plot(x, profit)
ax[1].set_title("Profit")

plt.show()
````

### **figsize**

When we have multiple subplots, the default figure can sometimes be too small. figsize lets us control the overall size of the figure.

Basic syntax

`fig, ax = plt.subplots(2, 1, figsize=(8, 6))`
```text
8 → figure width
6 → figure height
The units are inches

````python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]

sales = [10, 15, 12, 20, 18]
profit = [5, 8, 7, 12, 10]

fig, ax = plt.subplots(2, 1, figsize=(8, 6)) # this says our figure is 8 inches wide and 6 inches tall

ax[0].plot(x, sales)
ax[0].set_title("Sales")

ax[1].plot(x, profit)
ax[1].set_title("Profit")

plt.subplots_adjust(hspace=0.5) # hspace is used to give gap between horizontal plots while we can use wspace for vertical spacing

plt.show()
````

### **Multi row and multi columns**

2x2 figures

````python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]

sales = [10, 15, 12, 20, 18]
profit = [5, 8, 7, 12, 10]
expenses = [4, 6, 5, 8, 7]
customers = [20, 25, 22, 30, 28]

fig, ax = plt.subplots(2, 2, figsize=(8, 6))

ax[0, 0].plot(x, sales)
ax[0, 0].set_title("Sales")

ax[0, 1].plot(x, profit)
ax[0, 1].set_title("Profit")

ax[1, 0].plot(x, expenses)
ax[1, 0].set_title("Expenses")

ax[1, 1].plot(x, customers)
ax[1, 1].set_title("Customers")

plt.subplots_adjust(hspace=0.5, wspace=0.5)

plt.show()
````

### **sharex=True — same X-axis**

````python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]

sales = [100, 120, 90, 150, 180]
profit = [20, 30, 15, 40, 50]

fig, ax = plt.subplots(2, 1, figsize=(8, 6), sharex=True)

# First subplot
ax[0].plot(months, sales)
ax[0].set_title("Monthly Sales")
ax[0].set_ylabel("Sales")

# Second subplot
ax[1].plot(months, profit)
ax[1].set_title("Monthly Profit")
ax[1].set_ylabel("Profit")
ax[1].set_xlabel("Month")

plt.show()
````

### **sharey=True — same Y-axis**

````python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]

store_a = [100, 120, 90, 150, 180]
store_b = [80, 110, 100, 140, 170]

fig, ax = plt.subplots(1, 2, figsize=(10, 4), sharey=True)

# First subplot
ax[0].plot(months, store_a)
ax[0].set_title("Store A")
ax[0].set_xlabel("Month")
ax[0].set_ylabel("Sales")

# Second subplot
ax[1].plot(months, store_b)
ax[1].set_title("Store B")
ax[1].set_xlabel("Month")

plt.show()
````

### **tight_layout()**

When you have multiple subplots, labels and titles can sometimes overlap or get too close.

`tight_layout()` automatically adjusts the spacing between subplots.

````python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]

sales = [100, 120, 90, 150, 180]
profit = [20, 30, 15, 40, 50]

fig, ax = plt.subplots(2, 1, figsize=(8, 6))

ax[0].plot(months, sales)
ax[0].set_title("Monthly Sales")
ax[0].set_ylabel("Sales")

ax[1].plot(months, profit)
ax[1].set_title("Monthly Profit")
ax[1].set_xlabel("Month")
ax[1].set_ylabel("Profit")

plt.tight_layout()

plt.show()
````

### **fig.suptitle()**
```text
It helps to give one title to the overall figure.

````python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]

sales = [100, 120, 90, 150, 180]
profit = [20, 30, 15, 40, 50]

fig, ax = plt.subplots(2, 1, figsize=(8, 6))

ax[0].plot(months, sales)
ax[0].set_title("Sales")

ax[1].plot(months, profit)
ax[1].set_title("Profit")

fig.suptitle("Company Performance")

plt.tight_layout()

plt.show()
````

Create 4 subplots in a 2×2 grid using this data:

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [100, 120, 90, 150, 180]

Requirements:
```text
Create a 2 × 2 grid.
Set figsize=(10, 7).
Plot each dataset in its own subplot.
Give each subplot a title.
Add appropriate axis labels.
Give the entire figure the title "Business Performance".
Use tight_layout() while leaving space for the overall title.

````python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]

sales = [100, 120, 90, 150, 180]
profit = [20, 30, 15, 40, 50]
expenses = [80, 90, 75, 110, 130]
customers = [50, 60, 45, 70, 85]

# Create a 2 × 2 subplot grid
fig, ax = plt.subplots(2, 2, figsize=(10, 7))

# Sales
ax[0, 0].plot(months, sales)
ax[0, 0].set_title("Sales")
ax[0, 0].set_xlabel("Month")
ax[0, 0].set_ylabel("Sales")

# Profit
ax[0, 1].plot(months, profit)
ax[0, 1].set_title("Profit")
ax[0, 1].set_xlabel("Month")
ax[0, 1].set_ylabel("Profit")

# Expenses
ax[1, 0].plot(months, expenses)
ax[1, 0].set_title("Expenses")
ax[1, 0].set_xlabel("Month")
ax[1, 0].set_ylabel("Expenses")

# Customers
ax[1, 1].plot(months, customers)
ax[1, 1].set_title("Customers")
ax[1, 1].set_xlabel("Month")
ax[1, 1].set_ylabel("Customers")

# Overall figure title
fig.suptitle("Business Performance")

# Automatically adjust spacing and leave room for suptitle
plt.tight_layout(rect=[0, 0, 1, 0.95])

plt.show()
````
